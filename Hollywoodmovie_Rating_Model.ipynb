```python
df =pd.read_csv("/content/Data.csv")
df.head(2)
```

```python
x = df.drop(["Rating"] , axis=1)
y = df["Rating"]
```

```python
model = keras.Sequential([
    keras.layers.Dense(units=7, activation="relu" , input_shape=[7])
])
#model.add(keras.layers.Dense(units=7, activation="relu"))
model.add(keras.layers.Dense(units=1, activation="relu"))
```

```python
model.compile(optimizer="adam" , loss="mean_squared_error")
```

```python
model.summary()
```

```python
model.fit(x , y ,epochs=500)
```

```python
model0 = keras.Sequential([
    keras.layers.Dense(units=7, activation="relu" , input_shape=[7])
])
model0.add(keras.layers.Dense(units=7, activation="relu"))
model0.add(keras.layers.Dense(units=1, activation="relu"))
```

```python
model0.summary()
```
